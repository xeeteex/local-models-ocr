```python
from transformers import AutoModelForImageTextToText, AutoProcessor
import torch

model = AutoModelForImageTextToText.from_pretrained(
    "datalab-to/chandra-ocr-2", dtype=torch.bfloat16, device_map="auto"
)
model.eval()
model.processor = AutoProcessor.from_pretrained("datalab-to/chandra-ocr-2")
model.processor.tokenizer.padding_side = "left"
```

```python
from chandra.input import load_pdf_images
from chandra.model.hf import generate_hf
from chandra.model.schema import BatchInputItem
from chandra.output import parse_markdown

images = load_pdf_images("sample_input.pdf", None)

batch = [BatchInputItem(image=img, prompt_type="ocr_layout") for img in images]
results = generate_hf(batch, model)

for i, result in enumerate(results):
    print(f"--- Page {i+1} ---")
    print(parse_markdown(result.raw))
```
